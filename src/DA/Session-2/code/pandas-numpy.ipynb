```python
def convert_dtype(arr, new_dtype):
    """
    Convert the data type of a NumPy array to a new data type.

    Parameters:
    arr (numpy.ndarray): The input NumPy array.
    new_dtype (data-type): The desired data type to convert to.

    Returns:
    numpy.ndarray: A new NumPy array with the specified data type.
    """

    try:
        arr = np.asarray(arr)  
        return arr.astype(new_dtype)
    except Exception as e:
        raise TypeError(f"invalid target data type {new_dtype}") from e
    except Exception as e:
        raise ValueError(f"cannot convert array to {new_dtype}") from e    
```

```python
list1 = [1, 2, 3, 4, 5]
```

```python
convert_dtype(np.array(list1), np.float64)
```

```python
def check_dtype(arr, expected_dtype):
    """
    Check if the data type of a NumPy array matches the expected data type.

    Parameters:
    arr (numpy.ndarray): The input NumPy array.
    expected_dtype (data-type): The expected data type to check against.

    Returns:
    bool: True if the data type matches, False otherwise.
    """

    try:
        if not isinstance(arr, np.ndarray):
            raise TypeError("Input must be a NumPy array.")
        return arr.dtype == expected_dtype
    except Exception as e:
        raise TypeError(f"invalid target data type {expected_dtype}") from e
```

```python
def init_array(shape,value=0):
    """
    Initialize a NumPy array with a specified shape and fill it with a given value.

    Parameters:
    shape (tuple): The shape of the array to be created.
    value (scalar): The value to fill the array with.

    Returns:
    numpy.ndarray: A new NumPy array of the specified shape filled with the given value.
    """
    return np.full(shape, value)
```

```python
zeros = init_array((3, 4))
zeros
```

```python
def init_weights(n_features , method = 'zeros' , seed = None):
    """
    Initialize weights for a machine learning model based on the specified method.

    Parameters:
    n_features (int): The number of features (input dimensions).
    method (str): The method to initialize weights. Options are 'zeros', 'ones', or 'random'.

    Returns:
    numpy.ndarray: An array of initialized weights.
    """
    if not isinstance(n_features, int) or n_features <= 0:
        raise ValueError("n_features must be a positive integer.")
    if method == 'zeros':
        return np.zeros(n_features)
    elif method == 'ones':
        return np.ones(n_features)
    elif method == 'random':
        if seed is not None:
            np.random.seed(seed)
        return np.random.rand(n_features) *0.01

       
    else:
        raise ValueError("Invalid method. Choose from 'zeros', 'ones', or 'random'.")
```

```python
initial_weights = init_weights(10, method='random' , seed=20000)
initial_weights
```
