```python
# How many null values are there in each column?
null = df.isnull().sum()
ratio = null / len(df) * 100
ratio
```

```python
# Remove rows with missing values in the 'Embarked' column
df = df.dropna(subset=['Embarked'])
```

```python
df = df.drop("Cabin", axis=1 )
```

```python
median_age = df['Age'].median()
df['Age'] = df['Age'].fillna(median_age , inplace=True)
```

```python
pd.DataFrame({'null': df.isnull().sum(), 'ratio': df.isnull().sum() / len(df) * 100}).T
```

# Outliers

<img src="/home/mohamed_hisham/AI/ML-DEPI/src/DA/Session-3/code/Outiler-Boxplot.webp" alt="Outliers and IQR" width="1000">

```python
num_col = df.select_dtypes('number').columns
for col in num_col:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    print(f"Column: {col}, Q1: {Q1}, Q3: {Q3}")
    IQR = Q3 - Q1
    print(f"Column: {col}, IQR: {IQR}")
    lower_fence = Q1 - 1.5 * IQR
    print(f"Column: {col}, Lower Fence: {lower_fence}")
    upper_fence = Q3 + 1.5 * IQR
    print(f"Column: {col}, Upper Fence: {upper_fence}")
    
```
