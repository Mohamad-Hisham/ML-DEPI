# Gradient Descent for Linear Regression

Using the data from `Quiz.pdf`:

- $x = [1, 2, 3, 4]$
- $y = [2, 2.8, 3.6, 4.5]$
- Initial values: $w = 0$, $b = 0$
- Learning rate: $\alpha = 0.01$

The prediction is:

$$\hat{y}_i = w x_i + b$$

For mean squared error, the gradients are:

$$\frac{\partial L}{\partial w} = \frac{2}{n}\sum_{i=1}^{n}(\hat{y}_i-y_i)x_i$$

$$\frac{\partial L}{\partial b} = \frac{2}{n}\sum_{i=1}^{n}(\hat{y}_i-y_i)$$

The parameters are updated using:

$$w \leftarrow w - \alpha\frac{\partial L}{\partial w}, \qquad b \leftarrow b - \alpha\frac{\partial L}{\partial b}$$

```python
import numpy as np

x = np.array([1, 2, 3, 4], dtype=float)
y = np.array([2, 2.8, 3.6, 4.5], dtype=float)
alpha = 0.01
iterations = 10000
w = 0.0
b = 0.0


def gradients(w, b):
    predictions = w * x + b
    errors = predictions - y
    dw = (2 / len(x)) * np.sum(errors * x)
    db = (2 / len(x)) * np.sum(errors)
    loss = np.mean(errors ** 2)
    return dw, db, loss


# First iteration, matching the worked example in the PDF.
dw, db, loss = gradients(w, b)
first_w = w - alpha * dw
first_b = b - alpha * db

print(f"Initial gradients: dw = {dw:.4f}, db = {db:.4f}")
print(f"After first iteration: w = {first_w:.4f}, b = {first_b:.4f}")

w, b = first_w, first_b
history = [(1, w, b, loss)]

for iteration in range(2, iterations + 1):
    dw, db, loss = gradients(w, b)
    w -= alpha * dw
    b -= alpha * db
    history.append((iteration, w, b, loss))

final_dw, final_db, final_loss = gradients(w, b)
print(f"After {iterations} iterations:")
print(f"w = {w:.6f}")
print(f"b = {b:.6f}")
print(f"MSE = {final_loss:.8f}")
```
