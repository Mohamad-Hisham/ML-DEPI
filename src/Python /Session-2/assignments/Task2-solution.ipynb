```python
def analyze_email(email):
    # Input Validation
    if email.count("@") != 1:
        return "Invalid email"

    at_index = email.index("@")

    if "." not in email[at_index:]:
        return "Invalid email"

    # Extract Username
    username = email[:at_index]

    # Extract Domain
    domain_part = email[at_index + 1:]
    domain = domain_part[:domain_part.rfind(".")]

    # Check Domain Ending
    if email.endswith(".com"):
        domain_type = "Commercial Domain"
    elif email.endswith(".edu"):
        domain_type = "Educational Domain"
    else:
        domain_type = "Other Domain"

    return {
        "Username": username,
        "Domain": domain,
        "Domain Type": domain_type
    }


email = "Amit_ml@gmail.edu"

result = analyze_email(email)
print(result)
```
