```python
def train_model(model_name , *dataset , **hyperparameters):
    print(f"Training model: {model_name}")
    print("\nDatasets")
    for ds in dataset:
        print("_" , ds)
    print("\nHyperparameters")
    for key, value in hyperparameters.items():
        print(f"{key}: {value}")
```

```python
train_model(

    "fraid detection",
    "train.csv",
    "valid.csv",
    "test.csv",
    learning_rate = 0.01,
    batch_size = 32,
    epochs = 10,
    optimizer = "adam",
    early_stopping = True


)
```
