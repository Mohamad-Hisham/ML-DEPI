```python
def factorial(num : int):

    '''
    function to calculate the factorial of a number
    input: num (int): the number to calculate the factorial of
    output: int: the factorial of the input number
    '''
    if num == 0:
     return 0
    if num == 1:
     return 1

    return num * factorial(num - 1)
```

```python
def isPrime (num : int):
    '''
    function to check if a number is prime or not
    input: num (int): the number to check
    output: bool: True if the number is prime, False otherwise
    '''

    for i in range(2, num):
        if num % i == 0:
            return False
    return True
 
    
    
```

```python
def reverseWord(word : str):


    '''
    function to reverse a sentence into a list 

    args : string 

    output : reversed list of the words of the sentence

    eample :  reverseWord("Hello world") -->  ['world', 'Hello']


    '''

    lst = word.split(" ")
    print (lst)
    reversed_list = lst[::-1]
    return reversed_list
```

```python
def count_words_occurrences(sentence : str):
    list_of_words = sentence.split(" ")
    unique_Words = set(list_of_words)
    word_count = {}
    for word in unique_Words:
        word_count[word] = list_of_words.count(word)
    return word_count
    
```

```python
count_words_occurrences("Ahmed mohamed dexter mohamed")
```
